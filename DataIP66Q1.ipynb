##โหลด DATA

```python
df1 = pd.read_csv(m,dtype={'hcode': str, 'month_no': str})
df1
```

```python
df1.head()
```

```python
df1.isnull().sum()
```

```python
# 1. สร้างตารางจับคู่ (รหัสโรค -> ชื่อโรค) จากแถวที่ไม่เป็นค่าว่าง
pdx_dict = df1.dropna(subset=['pdx_name']).drop_duplicates(subset=['pdx']).set_index('pdx')['pdx_name'].to_dict()

# 2. อ่านรหัสโรคจากคอลัมน์ pdx แล้วเอาชื่อโรคมาเติมในช่อง pdx_name ที่ว่าง
df1['pdx_name'] = df1['pdx_name'].fillna(df1['pdx'].map(pdx_dict))
```

```python
df1.isnull().sum()
```

```python
print(f"Original number of rows: {len(df1)}")
df1.dropna(inplace=True)
print(f"Number of rows after dropping nulls: {len(df1)}")
```

```python
# แปลงรหัสเพศให้เป็นชื่อที่อ่านเข้าใจง่าย (1: ชาย, 2: หญิง, อื่นๆ: ไม่ระบุ)
df1['gender'] = df1['gender'].fillna(0)
gender_map = {1.0: 'ชาย', 2.0: 'หญิง', 0: 'ไม่ระบุ'}
df1['gender_label'] = df1['gender'].map(gender_map)
```

Top 5 ผู้ป่วยที่เข้ารับบริการส่วนมากเข้ามารับบริการเพราะโรคใด

```python
import plotly.express as px

# 1. เตรียมข้อมูลสรุปโรคและคำนวณสัดส่วนเปอร์เซ็นต์
df_disease = df1.groupby('pdx_name')['n_service'].sum().reset_index()
total_services = df_disease['n_service'].sum()
df_disease['percentage'] = (df_disease['n_service'] / total_services) * 100
df_disease = df_disease.sort_values(by='n_service', ascending=False)

# 2. ดึง 5 อันดับแรก
top5_disease = df_disease.head(5)

# 3. สร้างกราฟแท่งแนวนอนแสดงโรคที่เด่นที่สุด 5 อันดับแรกพร้อมเปอร์เซ็นต์สัดส่วน
fig = px.bar(
    top5_disease,
    x='n_service',
    y='pdx_name',
    orientation='h',
    text=top5_disease.apply(lambda r: f"{r['n_service']:,} ครั้ง ({r['percentage']:.2f}%)", axis=1),
    color='n_service',
    color_continuous_scale='Reds',
    title='5 อันดับโรค/อาการที่เด่นที่สุดและมีสัดส่วนผู้เข้าใช้บริการสูงที่สุด (Top Dominant Diseases)',
    labels={'pdx_name': 'ชื่อโรค / อาการ', 'n_service': 'จำนวนครั้งการให้บริการ'}
)

fig.update_layout(
    yaxis={'categoryorder': 'total ascending'},
    xaxis_title="จำนวนครั้งการเข้ารับบริการ (ครั้ง)",
    yaxis_title="",
    coloraxis_showscale=False,
    height=550,
    margin=dict(l=150, r=50, t=80, b=50)
)

fig.update_traces(textposition='inside', textfont=dict(color='white', size=11))
fig.show()
```

ใช้ Marimekko Chart สำหรับแสดงความสัมพันธ์ระหว่างโรคและเพศ

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go

# 1. กรองเฉพาะ Top 10 โรคที่มีผู้ใช้บริการสูงสุดเพื่อไม่ให้กราฟหนาแน่นเกินไป
top10_pdx = df1.groupby('pdx_name')['n_service'].sum().nlargest(10).index
df_filtered = df1[df1['pdx_name'].isin(top10_pdx)]

# 2. คำนวณสรุปข้อมูลแยกตาม โรค และ เพศ
df_grouped_pdx = df_filtered.groupby(['pdx_name', 'gender_label'])['n_service'].sum().unstack(fill_value=0)

# หาผลรวมของแต่ละโรค และสัดส่วนสะสมเพื่อเรียงลำดับจากมากไปน้อย
pdx_totals = df_grouped_pdx.sum(axis=1)
total_pdx_services = pdx_totals.sum()
pdx_shares = (pdx_totals / total_pdx_services).sort_values(ascending=False)

# เรียงลำดับข้อมูล
df_grouped_pdx = df_grouped_pdx.reindex(pdx_shares.index)
df_percentage_pdx = df_grouped_pdx.div(df_grouped_pdx.sum(axis=1), axis=0) * 100

# 3. คำนวณตำแหน่งแกน X และความกว้าง
widths_pdx = pdx_shares.values * 100
x_positions_pdx = np.cumsum(widths_pdx) - (widths_pdx / 2)

# 4. สร้าง Marimekko Chart
fig = go.Figure()
genders = ['ชาย', 'หญิง', 'ไม่ระบุ']
colors_map = {'ชาย': '#3182bd', 'หญิง': '#e377c2', 'ไม่ระบุ': '#7f7f7f'}

for g in genders:
    if g in df_percentage_pdx.columns:
        fig.add_trace(go.Bar(
            name=g,
            x=x_positions_pdx,
            y=df_percentage_pdx[g],
            width=widths_pdx,
            offset=-(widths_pdx / 2),
            marker=dict(color=colors_map.get(g, '#7f7f7f'), line=dict(color='white', width=1.5)),
            customdata=np.stack([
                df_grouped_pdx[g],
                [pdx_totals[p] for p in df_grouped_pdx.index],
                [pdx_shares[p] * 100 for p in df_grouped_pdx.index]
            ], axis=-1),
            hovertemplate=(
                "โรค: %{x_label}" +
                "เพศ: " + g + "" +
                "สัดส่วนเพศในโรคนี้: %{y:.1f}%" +
                "จำนวนการรักษาเฉพาะเพศ: %{customdata[0]:,} ครั้ง" +
                "จำนวนรวมของโรคนี้: %{customdata[1]:,} ครั้ง" +
                "ส่วนแบ่งของโรคนี้ใน Top 10: %{customdata[2]:.1f}%" +
                ""
            )
        ))

fig.update_layout(
    barmode='stack',
    title=dict(
        text='Marimekko Chart: สัดส่วนเพศและขนาดจำนวนการรักษาจำแนกตามกลุ่มโรค (Top 10)(ความกว้างสะท้อนสัดส่วนผู้เข้ารับรักษาของแต่ละโรค / ความสูงสะท้อนสัดส่วนเพศภายในโรคนั้น)',
        x=0.5
    ),
    xaxis=dict(
        title='สัดส่วนการใช้บริการสะสมของกลุ่มโรคหลัก (%)',
        tickvals=list(np.linspace(0, 100, 11)),
        ticktext=[f'{int(val)}%' for val in np.linspace(0, 100, 11)]
    ),
    yaxis=dict(
        title='สัดส่วนของแต่ละเพศ (%)',
        ticksuffix='%'
    ),
    legend_title='เพศ',
    height=750,
    width=1150,
    margin=dict(t=160, b=150, l=80, r=40)
)

# ปรับป้ายกำกับชื่อโรคด้านบนเพื่อป้องกันการซ้อนเบียด
for pdx, x_pos, width in zip(df_grouped_pdx.index, x_positions_pdx, widths_pdx):
    # ตัดคำสั้นลงเพื่อความกระชับและไม่เบียดเสียด
    short_name = pdx[:20] + '...' if len(pdx) > 20 else pdx
    fig.add_annotation(
        x=x_pos,
        y=105,
        text=f"{short_name}({width:.1f}%)",
        showarrow=False,
        font=dict(size=8.5, color='black'),
        align='center',
        textangle=-30 # เอียงองศาเพิ่มขึ้นเพื่อหลบเลี่ยงการซ้อนทับอย่างมีประสิทธิภาพ
    )

fig.show()
```

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go

# 1. เตรียมและคำนวณข้อมูลสำหรับ Marimekko Chart
# หาจำนวนรวมของบริการแบ่งตาม ภูมิภาค และ เพศ
df_grouped = df1.groupby(['region', 'gender_label'])['n_service'].sum().unstack(fill_value=0)

# หาผลรวมบริการรวมของแต่ละภูมิภาค และ สัดส่วนรวมของแต่ละภูมิภาคในประเทศ
region_totals = df_grouped.sum(axis=1)
total_services = region_totals.sum()
region_shares = (region_totals / total_services).sort_values(ascending=False)

# เรียงลำดับภูมิภาคตามขนาดจำนวนครั้งรับบริการจากมากไปน้อย
df_grouped = df_grouped.reindex(region_shares.index)

# แปลงข้อมูลจำนวนบริการในแต่ละภูมิภาคให้เป็นเปอร์เซ็นต์ (สัดส่วน 100% ภายในกลุ่ม)
df_percentage = df_grouped.div(df_grouped.sum(axis=1), axis=0) * 100

# 2. คำนวณตำแหน่งและสเกลความกว้างของแต่ละแท่งบนแกน X
widths = region_shares.values * 100  # แปลงสัดส่วนความกว้างเป็น % 0-100
x_positions = np.cumsum(widths) - (widths / 2)  # จุดกึ่งกลางของแต่ละแท่งสำหรับวางป้ายชื่อ

# 3. สร้างกราฟ Marimekko โดยใช้ go.Bar
fig = go.Figure()

# เพิ่มแท่งกราฟสำหรับแต่ละเพศ
genders = df_percentage.columns
colors = {'ชาย': '#3182bd', 'หญิง': '#e377c2', 'ไม่ระบุ': '#7f7f7f'}  # กำหนดโทนสี

for gender in genders:
    fig.add_trace(go.Bar(
        name=gender,
        x=x_positions,
        y=df_percentage[gender],
        width=widths, # กำหนดความกว้างของแท่งตามส่วนแบ่งของภูมิภาค
        offset=-(widths/2), # จัดแนวให้อยู่ตรงกลางตำแหน่ง x
        marker=dict(color=colors.get(gender, '#bcbd22'), line=dict(color='white', width=1.5)),
        customdata=np.stack([
            df_grouped[gender],
            [region_totals[r] for r in df_grouped.index],
            [region_shares[r]*100 for r in df_grouped.index]
        ], axis=-1),
        hovertemplate=(
            "ภูมิภาค: %{x_label}" +
            "เพศ: " + gender + "" +
            "สัดส่วนเพศในภูมิภาคนี้: %{y:.1f}%" +
            "จำนวนบริการเฉพาะเพศนี้: %{customdata[0]:,}" +
            "บริการทั้งหมดของภูมิภาคนี้: %{customdata[1]:,}" +
            "ส่วนแบ่งของภูมิภาคในประเทศ: %{customdata[2]:.1f}%" +
            ""
        )
    ))

# 4. ตั้งค่าหน้าตากราฟและแสดงผล
fig.update_layout(
    barmode='stack',
    title=dict(
        text='Marimekko Chart: สัดส่วนเพศและขนาดจำนวนผู้รับบริการจำแนกตามภูมิภาค(ความกว้างแท่งสะท้อนส่วนแบ่งผู้รับบริการของแต่ละภูมิภาค / ความสูงสะท้อนสัดส่วนเพศในภูมิภาคนั้น)',
        x=0.5
    ),
    xaxis=dict(
        title='สัดส่วนจำนวนบริการสะสมของภูมิภาค (%)',
        tickvals=list(np.linspace(0, 100, 11)),
        ticktext=[f'{int(val)}%' for val in np.linspace(0, 100, 11)]
    ),
    yaxis=dict(
        title='สัดส่วนของเพศ (%)',
        ticksuffix='%'
    ),
    legend_title='เพศ',
    height=700,
    width=1100,
    margin=dict(t=140, b=100, l=80, r=40)
)

# เพิ่มป้ายชื่อภูมิภาคที่แกน X ด้านบนตัวกราฟแบบเฉียงเพื่อกันตัวหนังสือเบียดกัน
for region, x_pos, width in zip(df_grouped.index, x_positions, widths):
    fig.add_annotation(
        x=x_pos,
        y=105,
        text=f"{region}({width:.1f}%)",
        showarrow=False,
        font=dict(size=10, color='black'),
        align='center',
        textangle=-15 # เอียงหลบเพื่อป้องกันตัวหนังสือซ้อนกัน
    )

fig.show()
```

```python
import plotly.express as px
import pandas as pd

# จัดกลุ่มข้อมูลตาม เดือน (month_no) และ เพศ (gender_label) เพื่อหาผลรวมจำนวนบริการ
df_monthly_gender = df1.groupby(['month_no', 'gender_label'])['n_service'].sum().reset_index()

# แปลงรหัสเดือนให้เป็นชื่อเดือนที่เข้าใจง่าย
month_map = {
    '202307': 'กรกฎาคม 2023',
    '202308': 'สิงหาคม 2023',
    '202309': 'กันยายน 2023'
}
df_monthly_gender['month_name'] = df_monthly_gender['month_no'].astype(str).map(month_map).fillna(df_monthly_gender['month_no'])

# เรียงลำดับข้อมูลตามรหัสเดือน
df_monthly_gender = df_monthly_gender.sort_values('month_no')

# สร้าง Rose Chart (Polar Bar Chart) โดยใช้ px.bar_polar และแก้ barmode เป็น 'stack'
fig = px.bar_polar(
    df_monthly_gender,
    r='n_service',
    theta='month_name',
    color='gender_label',
    color_discrete_map={'ชาย': '#3182bd', 'หญิง': '#e377c2', 'ไม่ระบุ': '#7f7f7f'},
    barmode='stack',
    title='จำนวนการเข้าใช้บริการจำแนกตามเพศในแต่ละเดือน',
    labels={'n_service': 'จำนวนครั้งการให้บริการ', 'month_name': 'เดือน', 'gender_label': 'เพศ'}
)

# ปรับแต่งการแสดงผลและป้ายข้อความ
fig.update_layout(
    polar=dict(
        angularaxis=dict(direction='clockwise', period=len(df_monthly_gender['month_name'].unique())),
        radialaxis=dict(ticksuffix='', angle=45)
    ),
    font=dict(size=12),
    height=600,
    width=800
)

fig.show()
```

ใช้ Stacked Area Chart สำหรับเปรียบเทียบแนวโน้มของโรคในแต่ละภูมิภาค

```python
import pandas as pd
import plotly.express as px

# 1. เตรียมข้อมูล
df_plot = df1[['region', 'pdx_name', 'n_service']].copy()

# 2. ทำความสะอาดข้อมูล
df_plot['n_service'] = pd.to_numeric(
    df_plot['n_service'], errors='coerce'
)

df_plot = df_plot.dropna(
    subset=['region', 'pdx_name', 'n_service']
)

# 3. หา Top 5 โรค
top_diseases = (
    df_plot.groupby('pdx_name')['n_service']
    .sum()
    .nlargest(5)
    .index
)

# 4. กรองข้อมูล Top 5 และรวมจำนวนบริการ
df_area = df_plot[
    df_plot['pdx_name'].isin(top_diseases)
].groupby(
    ['region', 'pdx_name'], as_index=False
)['n_service'].sum()

# 5. เรียงภูมิภาค
region_order = [
    'เหนือ', 'ตะวันออกเฉียงเหนือ', 'กลาง',
    'ตะวันออก', 'ตะวันตก', 'ใต้', 'อื่น ๆ'
]

# แปลงเป็นข้อความก่อนจัดลำดับ
df_area['region'] = df_area['region'].astype(str)

# เรียงตามลำดับที่กำหนด โดยภูมิภาคอื่นจะอยู่ท้าย
order = [r for r in region_order if r in df_area['region'].unique()]
order += [r for r in df_area['region'].unique() if r not in order]

print("จำนวนแถวที่ใช้วาดกราฟ:", len(df_area))
print(df_area.head(10))

# 6. สร้างกราฟแท่งแบบกลุ่ม
fig = px.bar(
    df_area,
    x='region',
    y='n_service',
    color='pdx_name',
    barmode='group',
    category_orders={'region': order},
    title='Top 5 โรคที่มีจำนวนการให้บริการสูงสุด จำแนกตามภูมิภาค',
    labels={
        'region': 'ภูมิภาค',
        'n_service': 'จำนวนครั้งการให้บริการ',
        'pdx_name': 'โรค/อาการ'
    },
    color_discrete_sequence=px.colors.qualitative.Safe,
    template='plotly_white'
)

fig.update_layout(
    xaxis_title='ภูมิภาค',
    yaxis_title='จำนวนครั้งการให้บริการ (ครั้ง)',
    legend=dict(
        orientation='h',
        yanchor='bottom',
        y=1.02,
        xanchor='center',
        x=0.5
    )
)

fig.show()
```

วิเคราะห์หาชื่อโรค/อาการที่โดดเด่นและมีผู้เข้าใช้บริการมากที่สุด

```python
import pandas as pd
import plotly.express as px

# 1. จัดกลุ่มและคำนวณผลรวมจำนวนครั้งการรับบริการ (n_service) แยกตามชื่อโรค (pdx_name)
df_disease = df1.groupby('pdx_name')['n_service'].sum().reset_index()

# 2. เรียงลำดับจากมากไปน้อย และคำนวณสัดส่วนสะสม (% Share)
total_all_services = df_disease['n_service'].sum()
df_disease['percentage'] = (df_disease['n_service'] / total_all_services) * 100
df_disease = df_disease.sort_values(by='n_service', ascending=False).reset_index(drop=True)

# ดึง Top 15 โรค/อาการที่เด่นที่สุด
top15_disease = df_disease.head(15)

# แสดงตารางวิเคราะห์
print("ตารางแสดง 15 อันดับโรค/อาการที่เด่นที่สุด (มีสัดส่วนผู้รับบริการสูงที่สุด):")
display(top15_disease.style.format({'n_service': '{:,}', 'percentage': '{:.2f}%'}))
```

```python
import plotly.graph_objects as go

# เรียงจากน้อย → มาก
df_lollipop = top15_disease.sort_values('n_service').copy()

# สีพาสเทล 15 สี
pastel_colors = [
    '#A8D8EA',  # ฟ้า
    '#F7B7D2',  # ชมพู
    '#B8E0D2',  # เขียวมิ้นต์
    '#FFD6A5',  # ส้มอ่อน
    '#CDB4DB',  # ม่วง
    '#FFF1A8',  # เหลือง
    '#AFCBFF',  # น้ำเงินอ่อน
    '#FFB7B2',  # แดงพาสเทล
    '#BDE0FE',  # ฟ้าอ่อน
    '#CDEAC0',  # เขียวอ่อน
    '#E2C2FF',  # ม่วงอ่อน
    '#FFD1DC',  # ชมพูอ่อน
    '#B5EAD7',  # เขียวมิ้นต์อ่อน
    '#FFDAC1',  # พีช
    '#D7BDE2'   # ม่วงลาเวนเดอร์
]

fig = go.Figure()

# วาดเส้นและจุดของแต่ละโรค
for i, (_, row) in enumerate(df_lollipop.iterrows()):

    # เส้น
    fig.add_trace(
        go.Scatter(
            x=[0, row['n_service']],
            y=[row['pdx_name'], row['pdx_name']],
            mode='lines',
            line=dict(
                color=pastel_colors[i],
                width=5
            ),
            showlegend=False,
            hoverinfo='skip'
        )
    )

    # จุด
    fig.add_trace(
        go.Scatter(
            x=[row['n_service']],
            y=[row['pdx_name']],
            mode='markers+text',
            marker=dict(
                size=15,
                color=pastel_colors[i],
                line=dict(
                    color='white',
                    width=2
                )
            ),
            text=[f"{row['n_service']:,}"],
            textposition='middle right',
            textfont=dict(
                size=11,
                color='#555555'
            ),
            showlegend=False,
            hovertemplate=(
                '%{y}'
                'จำนวนครั้งเข้ารับบริการ: %{x:,}'
                ''
            )
        )
    )

fig.update_layout(
    title=dict(
        text='15 อันดับโรค/อาการที่มีจำนวนครั้งเข้ารับบริการสูงที่สุด',
        x=0.5
    ),
    xaxis_title='จำนวนครั้งเข้ารับบริการ',
    yaxis_title='โรค/อาการ',

    height=700,

    margin=dict(
        l=300,
        r=100,
        t=90,
        b=60
    ),

    plot_bgcolor='white',
    paper_bgcolor='white',

    xaxis=dict(
        showgrid=True,
        gridcolor='#EEEEEE'
    ),

    yaxis=dict(
        showgrid=False
    )
)

fig.show()
```
