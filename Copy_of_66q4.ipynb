## ใช้ Treemap สำหรับแสดงสัดส่วนของโรคจำแนกตามภูมิภาค

```python
import plotly.express as px

# ป้องกัน KeyError: สร้างหรือแมปคอลัมน์ region ใหม่ให้ถูกต้องเรียบร้อยก่อน
if 'region_map' in globals():
    df4['region'] = df4['prov'].map(region_map)
else:
    from code_context import region_map
    df4['region'] = df4['prov'].map(region_map)

df4['region'] = df4['region'].fillna('ภาคกลาง')

# 1. จัดกลุ่มข้อมูลตามภูมิภาคและชื่อโรค
df_region_disease = df4.groupby(['region', 'pdx_name'], as_index=False)['n_service'].sum()

# 2. ดึง Top 5 โรคของแต่ละภูมิภาค
top_diseases_per_region = df_region_disease.sort_values(['region', 'n_service'], ascending=[True, False]).groupby('region').head(5).reset_index(drop=True)

# 3. สร้าง Treemap
fig = px.treemap(
    top_diseases_per_region,
    path=['region', 'pdx_name'],
    values='n_service',
    color='n_service',
    color_continuous_scale='Purples',  # เปลี่ยนเป็น Purples หรือ Plasma เพื่อความเข้มและชัดเจนขึ้น
    title='Treemap แสดงสัดส่วนโรคเด่น 5 อันดับแรกจำแนกตามภูมิภาค (ภูมิภาค -> ชื่อโรค)',
    labels={'region': 'ภูมิภาค', 'pdx_name': 'ชื่อโรค/อาการ', 'n_service': 'จำนวนครั้งการบริการ'}
)

# 4. ปรับแต่งการแสดงผลและสีข้อความให้ชัดเจน (แก้ปัญหาสีกลืน)
fig.update_traces(
    root_color="#f0f2f5",
    textinfo="label+value+percent parent",
    # เอา color="white" ออกเพื่อให้ Plotly ปรับสีตัวอักษร (ดำ/ขาว) ตามความสว่างพื้นหลังอัตโนมัติ
    textfont=dict(size=13, family="Tohoma, Sans-serif"),
    marker=dict(
        line=dict(color='white', width=2)  # เพิ่มเส้นขอบขาวแยกแต่ละบล็อกให้ดูง่ายขึ้น
    ),
    hovertemplate="""
<b>ระดับชั้น:</b> %{label}<br>
<b>จำนวนครั้งการให้บริการ:</b> %{value:,} ครั้ง<br>
<b>สัดส่วนจากกลุ่มหลัก:</b> %{percentParent:.2%}<br>
<b>สัดส่วนจากทั้งหมด:</b> %{percentRoot:.2%}
<extra></extra>
"""
)

fig.update_layout(
    margin=dict(t=60, l=10, r=10, b=10),
    height=750,
    font=dict(size=12),
    coloraxis_colorbar=dict(
        title="จำนวนครั้ง",
        thickness=15,
        len=0.8
    )
)

fig.show()
```

##ใช้ Marimekko Chart สำหรับแสดงความสัมพันธ์ระหว่างโรคและเพศ

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go

# 1. กรองเฉพาะ Top 10 โรคที่มีผู้ใช้บริการสูงสุดเพื่อไม่ให้กราฟหนาแน่นเกินไป
top10_pdx = df4.groupby('pdx_name')['n_service'].sum().nlargest(10).index
df_filtered = df4[df4['pdx_name'].isin(top10_pdx)]

# 2. คำนวณสรุปข้อมูลแยกตาม โรค และ เพศ
df_grouped_pdx = df_filtered.groupby(['pdx_name', 'gender_label'])['n_service'].sum().unstack(fill_value=0)

# หาผลรวมของแต่ละโรค และสัดส่วนสะสมเพื่อเรียงลำดับจากมากไปน้อย
pdx_totals = df_grouped_pdx.sum(axis=1)
total_pdx_services = pdx_totals.sum()
pdx_shares = (pdx_totals / total_pdx_services).sort_values(ascending=False)

# เรียงลำดับข้อมูล
df_grouped_pdx = df_grouped_pdx.reindex(pdx_shares.index)
df_percentage_pdx = df_grouped_pdx.div(df_grouped_pdx.sum(axis=1), axis=0) * 100

# 3. คำนวณตำแหน่งแกน X และความกว้าง
widths_pdx = pdx_shares.values * 100
x_positions_pdx = np.cumsum(widths_pdx) - (widths_pdx / 2)

# 4. สร้าง Marimekko Chart
fig = go.Figure()
genders = ['ชาย', 'หญิง', 'ไม่ระบุ']
colors_map = {'ชาย': '#3182bd', 'หญิง': '#e377c2', 'ไม่ระบุ': '#7f7f7f'}

for g in genders:
    if g in df_percentage_pdx.columns:
        fig.add_trace(go.Bar(
            name=g,
            x=x_positions_pdx,
            y=df_percentage_pdx[g],
            width=widths_pdx,
            offset=-(widths_pdx / 2),
            marker=dict(color=colors_map.get(g, '#7f7f7f'), line=dict(color='white', width=1.5)),
            customdata=np.stack([
                df_grouped_pdx[g],
                [pdx_totals[p] for p in df_grouped_pdx.index],
                [pdx_shares[p] * 100 for p in df_grouped_pdx.index]
            ], axis=-1),
            hovertemplate=(
                "<b>โรค:</b> %{x_label}<br>" +
                "<b>เพศ:</b> " + g + "<br>" +
                "<b>สัดส่วนเพศในโรคนี้:</b> %{y:.1f}%<br>" +
                "<b>จำนวนการรักษาเฉพาะเพศ:</b> %{customdata[0]:,} ครั้ง<br>" +
                "<b>จำนวนรวมของโรคนี้:</b> %{customdata[1]:,} ครั้ง<br>" +
                "<b>ส่วนแบ่งของโรคนี้ใน Top 10:</b> %{customdata[2]:.1f}%<br>" +
                "<extra></extra>"
            )
        ))

fig.update_layout(
    barmode='stack',
    title=dict(
        text='Marimekko Chart: สัดส่วนเพศและขนาดจำนวนการรักษาจำแนกตามกลุ่มโรค (Top 10)<br><sup>(ความกว้างสะท้อนสัดส่วนผู้เข้ารับรักษาของแต่ละโรค / ความสูงสะท้อนสัดส่วนเพศภายในโรคนั้น)</sup>',
        x=0.5
    ),
    xaxis=dict(
        title='สัดส่วนการใช้บริการสะสมของกลุ่มโรคหลัก (%)',
        tickvals=list(np.linspace(0, 100, 11)),
        ticktext=[f'{int(val)}%' for val in np.linspace(0, 100, 11)]
    ),
    yaxis=dict(
        title='สัดส่วนของแต่ละเพศ (%)',
        ticksuffix='%'
    ),
    legend_title='เพศ',
    height=750,
    width=1150,
    margin=dict(t=160, b=150, l=80, r=40)
)

# ปรับป้ายกำกับชื่อโรคด้านบนเพื่อป้องกันการซ้อนเบียด
for pdx, x_pos, width in zip(df_grouped_pdx.index, x_positions_pdx, widths_pdx):
    # ตัดคำสั้นลงเพื่อความกระชับและไม่เบียดเสียด
    short_name = pdx[:20] + '...' if len(pdx) > 20 else pdx
    fig.add_annotation(
        x=x_pos,
        y=105,
        text=f"<b>{short_name}</b><br>({width:.1f}%)",
        showarrow=False,
        font=dict(size=8.5, color='black'),
        align='center',
        textangle=-30 # เอียงองศาเพิ่มขึ้นเพื่อหลบเลี่ยงการซ้อนทับอย่างมีประสิทธิภาพ
    )

fig.show()
```

##เพศชายหญิง

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go

# 1. เตรียมและคำนวณข้อมูลสำหรับ Marimekko Chart
# หาจำนวนรวมของบริการแบ่งตาม ภูมิภาค และ เพศ
df_grouped = df4.groupby(['region', 'gender_label'])['n_service'].sum().unstack(fill_value=0)

# หาผลรวมบริการรวมของแต่ละภูมิภาค และ สัดส่วนรวมของแต่ละภูมิภาคในประเทศ
region_totals = df_grouped.sum(axis=1)
total_services = region_totals.sum()
region_shares = (region_totals / total_services).sort_values(ascending=False)

# เรียงลำดับภูมิภาคตามขนาดจำนวนครั้งรับบริการจากมากไปน้อย
df_grouped = df_grouped.reindex(region_shares.index)

# แปลงข้อมูลจำนวนบริการในแต่ละภูมิภาคให้เป็นเปอร์เซ็นต์ (สัดส่วน 100% ภายในกลุ่ม)
df_percentage = df_grouped.div(df_grouped.sum(axis=1), axis=0) * 100

# 2. คำนวณตำแหน่งและสเกลความกว้างของแต่ละแท่งบนแกน X
widths = region_shares.values * 100  # แปลงสัดส่วนความกว้างเป็น % 0-100
x_positions = np.cumsum(widths) - (widths / 2)  # จุดกึ่งกลางของแต่ละแท่งสำหรับวางป้ายชื่อ

# 3. สร้างกราฟ Marimekko โดยใช้ go.Bar
fig = go.Figure()

# เพิ่มแท่งกราฟสำหรับแต่ละเพศ
genders = df_percentage.columns
colors = {'ชาย': '#3182bd', 'หญิง': '#e377c2', 'ไม่ระบุ': '#7f7f7f'}  # กำหนดโทนสี

for gender in genders:
    fig.add_trace(go.Bar(
        name=gender,
        x=x_positions,
        y=df_percentage[gender],
        width=widths, # กำหนดความกว้างของแท่งตามส่วนแบ่งของภูมิภาค
        offset=-(widths/2), # จัดแนวให้อยู่ตรงกลางตำแหน่ง x
        marker=dict(color=colors.get(gender, '#bcbd22'), line=dict(color='white', width=1.5)),
        customdata=np.stack([
            df_grouped[gender],
            [region_totals[r] for r in df_grouped.index],
            [region_shares[r]*100 for r in df_grouped.index]
        ], axis=-1),
        hovertemplate=(
            "<b>ภูมิภาค:</b> %{x_label}<br>" +
            "<b>เพศ:</b> " + gender + "<br>" +
            "<b>สัดส่วนเพศในภูมิภาคนี้:</b> %{y:.1f}%<br>" +
            "<b>จำนวนบริการเฉพาะเพศนี้:</b> %{customdata[0]:,}<br>" +
            "<b>บริการทั้งหมดของภูมิภาคนี้:</b> %{customdata[1]:,}<br>" +
            "<b>ส่วนแบ่งของภูมิภาคในประเทศ:</b> %{customdata[2]:.1f}%<br>" +
            "<extra></extra>"
        )
    ))

# 4. ตั้งค่าหน้าตากราฟและแสดงผล
fig.update_layout(
    barmode='stack',
    title=dict(
        text='Marimekko Chart: สัดส่วนเพศและขนาดจำนวนผู้รับบริการจำแนกตามภูมิภาค<br><sup>(ความกว้างแท่งสะท้อนส่วนแบ่งผู้รับบริการของแต่ละภูมิภาค / ความสูงสะท้อนสัดส่วนเพศในภูมิภาคนั้น)</sup>',
        x=0.5
    ),
    xaxis=dict(
        title='สัดส่วนจำนวนบริการสะสมของภูมิภาค (%)',
        tickvals=list(np.linspace(0, 100, 11)),
        ticktext=[f'{int(val)}%' for val in np.linspace(0, 100, 11)]
    ),
    yaxis=dict(
        title='สัดส่วนของเพศ (%)',
        ticksuffix='%'
    ),
    legend_title='เพศ',
    height=700,
    width=1100,
    margin=dict(t=140, b=100, l=80, r=40)
)

# เพิ่มป้ายชื่อภูมิภาคที่แกน X ด้านบนตัวกราฟแบบเฉียงเพื่อกันตัวหนังสือเบียดกัน
for region, x_pos, width in zip(df_grouped.index, x_positions, widths):
    fig.add_annotation(
        x=x_pos,
        y=105,
        text=f"<b>{region}</b><br>({width:.1f}%)",
        showarrow=False,
        font=dict(size=10, color='black'),
        align='center',
        textangle=-15 # เอียงหลบเพื่อป้องกันตัวหนังสือซ้อนกัน
    )

fig.show()
```

##ใช้ Nightingale Rose Chart สำหรับแสดงการกระจายของผู้ป่วยในแต่ละเดือน

```python
import plotly.express as px
import pandas as pd

# 1. จัดกลุ่มข้อมูลตาม เดือน (month_no) และ เพศ (gender_label) เพื่อหาผลรวมจำนวนบริการ
df_monthly_gender = df4.groupby(['month_no', 'gender_label'])['n_service'].sum().reset_index()

# 2. แปลงรหัสเดือนให้เป็นชื่อเดือนที่เข้าใจง่าย
month_map = {
    '202307': 'กรกฎาคม 2023',
    '202308': 'สิงหาคม 2023',
    '202309': 'กันยายน 2023'
}
df_monthly_gender['month_name'] = df_monthly_gender['month_no'].astype(str).map(month_map).fillna(df_monthly_gender['month_no'])

# เรียงลำดับข้อมูลตามรหัสเดือน
df_monthly_gender = df_monthly_gender.sort_values('month_no')

# 3. สร้าง Rose Chart (Polar Bar Chart) โดยใช้ px.bar_polar และแก้ barmode เป็น 'stack'
fig = px.bar_polar(
    df_monthly_gender,
    r='n_service',
    theta='month_name',
    color='gender_label',
    color_discrete_map={'ชาย': '#3182bd', 'หญิง': '#e377c2', 'ไม่ระบุ': '#7f7f7f'},
    barmode='stack',
    title='จำนวนการเข้าใช้บริการจำแนกตามเพศในแต่ละเดือน',
    labels={'n_service': 'จำนวนครั้งการให้บริการ', 'month_name': 'เดือน', 'gender_label': 'เพศ'}
)

# ปรับแต่งการแสดงผลและป้ายข้อความ
fig.update_layout(
    polar=dict(
        angularaxis=dict(direction='clockwise', period=len(df_monthly_gender['month_name'].unique())),
        radialaxis=dict(ticksuffix='', angle=45)
    ),
    font=dict(size=12),
    height=600,
    width=800
)

fig.show()
```

##ใช้ Stacked Area Chart สำหรับเปรียบเทียบแนวโน้มของโรคในแต่ละภูมิภาค

```python
import plotly.express as px
import pandas as pd

# 1. หา Top 5 โรคที่มีจำนวนการบริการสูงที่สุดในภาพรวม
top_diseases = df4.groupby('pdx_name')['n_service'].sum().nlargest(5).index

# 2. กรองข้อมูลเฉพาะ Top 5 โรคนี้ และจัดกลุ่มตามภูมิภาคและโรค
df_filtered = df4[df4['pdx_name'].isin(top_diseases)]
df_area = df_filtered.groupby(['region', 'pdx_name'], as_index=False)['n_service'].sum()

# เรียงลำดับภูมิภาคเพื่อให้แกน X เรียงตัวสวยงาม
region_order = ['เหนือ', 'ตะวันออกเฉียงเหนือ', 'กลาง', 'ตะวันออก', 'ตะวันตก', 'ใต้', 'อื่น ๆ']
df_area['region'] = pd.Categorical(df_area['region'], categories=region_order, ordered=True)
df_area = df_area.sort_values('region')

# 3. สร้าง Stacked Area Chart
fig = px.area(
    df_area,
    x='region',
    y='n_service',
    color='pdx_name',
    title='Stacked Area Chart: เปรียบเทียบจำนวนการบริการของโรคหลักจำแนกตามภูมิภาค',
    labels={'region': 'ภูมิภาค', 'n_service': 'จำนวนครั้งการให้บริการ', 'pdx_name': 'โรค/อาการ'},
    color_discrete_sequence=px.colors.qualitative.Safe
)

fig.update_layout(
    xaxis_title='ภูมิภาค',
    yaxis_title='จำนวนครั้งการให้บริการ (ครั้ง)',
    hovermode='x unified',
    legend=dict(orientation='h', yanchor='bottom', y=1.02, xanchor='right', x=1)
)

fig.show()
```
